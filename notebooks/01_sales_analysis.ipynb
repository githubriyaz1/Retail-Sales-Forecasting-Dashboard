```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go

pd.set_option("display.max_columns", None)

df = pd.read_csv(
    "../data/processed/retail_sales_clean.csv"
)

df["Order_Date"] = pd.to_datetime(
    df["Order_Date"]
)

print("Dataset shape:", df.shape)

df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
category_sales = (
    df.groupby("Category")["Revenue"]
    .sum()
    .sort_values(ascending=False)
)

category_sales
```

```python
fig = px.bar(
    category_sales,
    x=category_sales.index,
    y=category_sales.values,
    title="Revenue by Product Category",
    labels={
        "x": "Category",
        "y": "Revenue"
    }
)

fig.show()
```

```python
region_sales = (
    df.groupby("Region")["Revenue"]
    .sum()
    .sort_values(ascending=False)
)

region_sales
```

```python
fig = px.bar(
    region_sales,
    x=region_sales.index,
    y=region_sales.values,
    title="Revenue by Region",
    labels={
        "x": "Region",
        "y": "Revenue"
    }
)

fig.show()
```

```python
monthly_sales = (
    df.groupby(df["Order_Date"].dt.to_period("M"))["Revenue"]
    .sum()
    .reset_index()
)

monthly_sales.head()
```

```python
monthly_sales["Order_Date"] = monthly_sales["Order_Date"].dt.to_timestamp()

fig = px.line(
    monthly_sales,
    x="Order_Date",
    y="Revenue",
    markers=True,
    title="Monthly Revenue Trend",
    labels={
        "Order_Date": "Month",
        "Revenue": "Revenue"
    }
)

fig.show()
```

```python
# Monthly Revenue Trend

monthly_sales = (
    df.groupby(
        df["Order_Date"].dt.to_period("M")
    )["Revenue"]
    .sum()
    .reset_index()
)

monthly_sales["Order_Date"] = (
    monthly_sales["Order_Date"]
    .dt.to_timestamp()
)

monthly_sales.head()
```

```python
top_products = (
    df.groupby("Product_Name")["Revenue"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_products
```

```python
fig = px.bar(
    top_products.sort_values(),
    orientation="h",
    title="Top 10 Products by Revenue",
    labels={
        "value": "Revenue",
        "Product_Name": "Product"
    }
)

fig.show()
```

```python
customer_sales = (
    df.groupby("Customer_Type")
    .agg(
        Revenue=("Revenue", "sum"),
        Profit=("Profit", "sum"),
        Orders=("Order_ID", "nunique")
    )
    .reset_index()
)

customer_sales
```

```python
fig = px.pie(
    customer_sales,
    names="Customer_Type",
    values="Revenue",
    title="Revenue Contribution by Customer Type"
)

fig.show()
```

```python
payment_sales = (
    df.groupby("Payment_Method")["Revenue"]
    .sum()
    .sort_values(ascending=False)
)

payment_sales
```

```python
fig = px.bar(
    payment_sales,
    x=payment_sales.index,
    y=payment_sales.values,
    title="Revenue by Payment Method",
    labels={
        "x": "Payment Method",
        "y": "Revenue"
    }
)

fig.show()
```

```python
discount_analysis = (
    df.groupby("Discount")
    .agg(
        Revenue=("Revenue", "sum"),
        Profit=("Profit", "sum"),
        Profit_Margin=("Profit_Margin", "mean")
    )
    .reset_index()
)

discount_analysis
```

```python
fig = px.line(
    discount_analysis,
    x="Discount",
    y="Profit_Margin",
    markers=True,
    title="Discount vs Average Profit Margin",
    labels={
        "Discount": "Discount (%)",
        "Profit_Margin": "Profit Margin (%)"
    }
)

fig.show()
```

```python
yearly_sales = (
    df.groupby("Year")["Revenue"]
    .sum()
    .reset_index()
)

yearly_sales["Growth_Percentage"] = (
    yearly_sales["Revenue"]
    .pct_change()
    * 100
)

yearly_sales
```

```python
fig = px.bar(
    yearly_sales,
    x="Year",
    y="Revenue",
    title="Yearly Revenue",
    text_auto=".2s",
    labels={
        "Year": "Year",
        "Revenue": "Revenue"
    }
)

fig.show()
```
